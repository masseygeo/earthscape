# **Imports**

```python
from earthscape.utils.constants import DATASET_DIR, STR_VERSION, SG_MAPPING, MODALITIES
from earthscape.data.patches import patches_get_stats

import os
import glob
import json
import pandas as pd
import geopandas as gpd
```

# **Full EarthScape Dataset**

## *Areas, Patch Locations, & Class Mappings*

```python
######################################################################
# Compile local dataset area, label, patch files into single global files
######################################################################

# paths to local area/patch/mapping files
local_dataset_dirs = [os.path.join(DATASET_DIR, d) for d in os.listdir(DATASET_DIR)\
                      if os.path.isdir(os.path.join(DATASET_DIR, d))]


# get all local area/patch/mapping paths...
area_paths = [glob.glob(f"{ldd}/*areas.csv")[0] for ldd in local_dataset_dirs]
patch_paths = [glob.glob(f"{ldd}/*patches.geojson")[0] for ldd in local_dataset_dirs]


# create global class-area proportion per patch csv...
areas = []
for path in area_paths:
    df = pd.read_csv(path)
    areas.append(df)
df_areas = pd.concat(areas, ignore_index=True)
areas_out = f"{DATASET_DIR}/es_{STR_VERSION}_areas.csv"
df_areas.to_csv(areas_out, index=False)


# create global patches geospatial vector polygon GeoJSON...
patches = []
for path in patch_paths:
    gdf = gpd.read_file(path)
    patches.append(gdf)
gdf_patches = pd.concat(patches, ignore_index=True)
patches_out = f"{DATASET_DIR}/es_{STR_VERSION}_patches.geojson"
gdf_patches.to_file(patches_out, driver='GeoJSON', index=False)


# create global class mapping json...
mapping_out = f"{DATASET_DIR}/es_{STR_VERSION}_class_map.json"
with open(mapping_out, "w") as f:
    json.dump(SG_MAPPING, f, indent=4)
```

## *Patch Statistics*

```python
########################################
# Calculate Global Dataset Statistics
########################################

# calculate dataset mean, sample sd
df_stats = patches_get_stats(data_dir=DATASET_DIR, modalities=MODALITIES, patch_ids=None)

# save as csv
stats_out = f"{DATASET_DIR}/es_{STR_VERSION}_stats.csv"
df_stats.to_csv(stats_out, index=True)

# display df head
df_stats.head()
```
