# **Imports**

```python
from earthscape.constants import DATASET_DIR, STR_VERSION, MODALITIES
from earthscape.data import patches_get_stats, create_release_metadata

import os
import glob
import json
import pandas as pd
import geopandas as gpd
```

# **Areas, Labels, & Patches**

```python
######################################################################
# Compile local dataset area, label, patch files into single global files
######################################################################

# paths to local area/patch/mapping files
local_dataset_dirs = [os.path.join(DATASET_DIR, d) for d in os.listdir(DATASET_DIR)\
                      if os.path.isdir(os.path.join(DATASET_DIR, d))  if 'smoke' not in d]


# get all local area/patch/mapping paths...
area_paths = [glob.glob(f"{ldd}/*areas.csv")[0] for ldd in local_dataset_dirs]
label_paths = [glob.glob(f"{ldd}/*labels.csv")[0] for ldd in local_dataset_dirs]
patch_paths = [glob.glob(f"{ldd}/*patches.geojson")[0] for ldd in local_dataset_dirs]


# create global class-area proportion per patch csv...
areas = []
for path in area_paths:
    df = pd.read_csv(path)
    areas.append(df)
df_areas = pd.concat(areas, ignore_index=True)
areas_out = f"{DATASET_DIR}/{STR_VERSION}_areas.csv"
df_areas.to_csv(areas_out, index=False)


# create global class-area proportion per patch csv...
labels = []
for path in label_paths:
    df = pd.read_csv(path)
    labels.append(df)
df_labels = pd.concat(labels, ignore_index=True)
labels_out = f"{DATASET_DIR}/{STR_VERSION}_labels.csv"
df_labels.to_csv(labels_out, index=False)


# create global patches geospatial vector polygon GeoJSON...
patches = []
for path in patch_paths:
    gdf = gpd.read_file(path)
    patches.append(gdf)
gdf_patches = pd.concat(patches, ignore_index=True)
patches_out = f"{DATASET_DIR}/{STR_VERSION}_patches.geojson"
gdf_patches.to_file(patches_out, driver='GeoJSON', index=False)
```

# **Patch Statistics**

```python
########################################
# Calculate Global Dataset Statistics
########################################

# calculate dataset mean, sample sd
df_stats = patches_get_stats(data_dir=DATASET_DIR, modalities=MODALITIES, patch_ids=None)

# save as csv
stats_out = f"{DATASET_DIR}/es{STR_VERSION}_stats.csv"
df_stats.to_csv(stats_out, index=True)

# display df head
df_stats.head()
```

# **Metadata**

```python
##################################################
# Create Dataset Release Metadata from Subsets
##################################################

# paths to subset metadata files
metadata_dirs = [os.path.join(DATASET_DIR, d) for d in os.listdir(DATASET_DIR)\
                 if os.path.isdir(os.path.join(DATASET_DIR, d))]


# open each file and create dict...
combined = {}
for dir_path in metadata_dirs:
    name = os.path.basename(dir_path)
    metadata_path = glob.glob(os.path.join(dir_path, '*metadata.json'))[0]
    with open(metadata_path, "r") as f:
        data = json.load(f)
    combined[name] = data


# combine metadata files into singular global metadata & save as JSON...
merged = create_release_metadata(combined)
output_path = os.path.join(DATASET_DIR, f"{STR_VERSION}_metadata.json")
with open(output_path, "w") as f:
    json.dump(merged, f, indent=4)
```
